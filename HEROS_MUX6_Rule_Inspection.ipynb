# HEROS MUX6 Rule Inspection

This notebook loads saved `MUX6` HEROS rule artifacts if they already exist. If they do not exist yet, it retrains the `MUX6` HEROS model once using the experiment configuration and saves reusable inspection artifacts.

Saved artifacts:
- `summary.json`
- `top_model_rules.csv`
- `model_population.csv`
- `rule_population.csv`

```python
from pathlib import Path
import json

import pandas as pd
from IPython.display import display

from evaluation.experiments.heros_llm.pretrained_mux6_rules import (
    ensure_mux6_rule_artifacts,
    load_mux6_rule_summary,
)
```

```python
artifacts = ensure_mux6_rule_artifacts(force_retrain=False)
summary = load_mux6_rule_summary(artifacts.output_dir)

print('Artifact directory:', artifacts.output_dir)
display(pd.Series(summary, name='MUX6 HEROS Summary'))
```

```python
top_model_rules = pd.read_csv(artifacts.top_model_rules_path)
print('Top model rules:', len(top_model_rules))
display(top_model_rules)
```

```python
model_population = pd.read_csv(artifacts.model_population_path)
print('Model population size:', len(model_population))
display(model_population)
```

```python
rule_population = pd.read_csv(artifacts.rule_population_path)
print('Full learned rule population size:', len(rule_population))
display(rule_population)
```
